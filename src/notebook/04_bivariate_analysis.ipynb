#INSIGHTS
- Higher household income is associated with higher car price.
- Older customers tend to buy more expensive vehicles.
- Hatchback buyers are generally younger and lower-income.
- Sedan buyers sit in the middle on both age and income.
- SUV buyers are older and wealthier on average.
- The relationship between income and car choice is strong enough to support customer segmentation.
- The relationship is associative, not causal.
